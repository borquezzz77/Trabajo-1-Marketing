```python
import pandas as pd
```

```python
df =  pd.read_csv('data_arriendos_bicicletas.csv')
```

```python
df
```

```python
df = df[['humedad', 'velocidad_viento']]
df
```

```python
filas_con_nan = df[df.isna().any(axis=1)]
cantidad_borradas = len(filas_con_nan)

df = df.dropna()

print("Cantidad de filas borradas:", cantidad_borradas)
print("Filas que se borraron:\n", filas_con_nan)

df
```

```python
criterio_filtro = (df['humedad'] >= 0) & (df['humedad'] <= 100) & (df['velocidad_viento'] >= 0) & (df['velocidad_viento'] <= 250)

filas_eliminadas = df[~criterio_filtro].copy()
filas_eliminadas['Fila_Excel'] = filas_eliminadas.index + 2

df = df[criterio_filtro]

print("Cantidad de filas eliminadas por inconsistencia física:", len(filas_eliminadas))
print("Detalle de filas eliminadas:")
filas_eliminadas
```
