```python
import pandas as pd

# 1. Cargar el set de datos asignado al Mercado 2 [cite: 56]
df = pd.read_csv(r"C:\Users\mambe\OneDrive\Escritorio\MARKETING\trabajo semestral\data_arriendos_bicicletas.csv", encoding='latin-1', sep=',')

# 2. Filtrar para mantener solo valores entre -25 y 50 grados 
# La condición mantiene las filas donde: -25 <= temperatura <= 50
df_filtrado = df[(df['temperatura'] >= -25) & (df['temperatura'] <= 50)].copy()

# 3. Verificar cuántas filas se eliminaron
eliminadas = len(df) - len(df_filtrado)
print(f"Se eliminaron {eliminadas} filas con temperaturas fuera de rango.")

# 4. Guardar el archivo limpio para el análisis posterior [cite: 19]
df_filtrado.to_csv('data_bicicletas_limpio.csv', index=False)

# Mostrar un resumen estadístico de la nueva columna para confirmar
print(df_filtrado['temperatura'].describe())

df_filtrado.drop(columns=["sensacion_termica"], inplace=True)
```
